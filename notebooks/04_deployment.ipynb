# 04 — Deployment

## Arabic Character-Level Language Model

In this notebook, we deploy the Arabic character-level N-gram language model as an interactive Streamlit application.

The goal is to transform the language modeling project into a simple web application that allows users to generate Arabic text interactively.

# 1. Project Architecture

The complete project pipeline is:
```
Raw Arabic Text
        ↓
Data Cleaning
        ↓
Character Tokenization
        ↓
Vocabulary
        ↓
Encoding / Decoding
        ↓
N-gram Language Model
        ↓
Text Generation
        ↓
Evaluation
        ↓
Streamlit Application
        ↓
Deployment
```

````python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent))

from src.preprocessing import load_text_file, clean_dataset
from src.tokenizer import EnhancedTokenizer
from src.ngram import CharacterNGramModel
from src.generator import generate_text
````

````python
from pathlib import Path

dataset_path = Path(
    "../data/raw/arabic_sample.txt"
)

dataset = load_text_file(
    dataset_path
)

dataset = clean_dataset(
    dataset
)

print(
    f"Number of samples: {len(dataset)}"
)

print("\nFirst samples:")

for text in dataset[:5]:
    print(text)
````

````python
tokenizer = EnhancedTokenizer(
    dataset
)

model = CharacterNGramModel(
    n=4
)

model.fit(
    dataset,
    tokenizer,
)

print(model)
print(
    f"Vocabulary size: "
    f"{tokenizer.vocabulary_size}"
)
print(
    f"Number of contexts: "
    f"{len(model)}"
)
````

````python
prompt = "يوم واحد"

generated_text = generate_text(
    start_prompt=prompt,
    n_tokens=50,
    model=model,
    tokenizer=tokenizer,
    sampling_mode="random",
    seed=42,
)

print(
    "Prompt:"
)

print(prompt)

print(
    "\nGenerated text:"
)

print(generated_text)
````

````python
generated_text_greedy = generate_text(
    start_prompt=prompt,
    n_tokens=50,
    model=model,
    tokenizer=tokenizer,
    sampling_mode="greedy",
)

print(generated_text_greedy)
````

````python
from src.evaluation import evaluate_model

metrics = evaluate_model(
    model,
    dataset,
    tokenizer,
)

print(
    f"Cross Entropy: "
    f"{metrics['cross_entropy']:.4f}"
)

print(
    f"Perplexity: "
    f"{metrics['perplexity']:.4f}"
)
````

## Evaluation

The model is evaluated using cross entropy and perplexity.

Lower values indicate better predictive performance.

For the final experiment, the official evaluation should be performed on a held-out test set rather than the training dataset.

# 2. Streamlit Deployment

The trained model is integrated into a Streamlit application.

The application provides:

- Arabic text input
- N-gram model selection
- Random generation
- Greedy generation
- Vocabulary information
- Model context information
- Cross entropy
- Perplexity

The application entry point is:

```text
app.py
```
run in terminal:
```
streamlit run app.py
```

# 3. Application Workflow
```
The Streamlit application follows this workflow:

User enters prompt
        ↓
Tokenizer converts prompt into characters
        ↓
N-gram model finds the current context
        ↓
Model calculates possible next characters
        ↓
Random or greedy sampling selects next character
        ↓
New character is added to generated text
        ↓
Process repeats
        ↓
Generated Arabic text is displayed
