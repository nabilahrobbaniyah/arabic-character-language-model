```python
from src.tokenizer import ArabicCharacterTokenizer


dataset = [
    "الشمس طلعت",
    "يوم جميل",
    "السماء صافية",
]


tokenizer = ArabicCharacterTokenizer()

tokenizer.fit(dataset)


print("Vocabulary:")
print(tokenizer.vocabulary)

print()
print("Vocabulary size:")
print(tokenizer.vocabulary_size)


text = "الشمس طلعت"

tokens = tokenizer.character_tokenize(text)

print()
print("Original text:")
print(text)

print()
print("Character tokens:")
print(tokens)


encoded = tokenizer.encode(text)

print()
print("Encoded:")
print(encoded)


decoded = tokenizer.decode(encoded)

print()
print("Decoded:")
print(decoded)


print()
print("Round-trip correct:")
print(text == decoded)
```
